```python
 import numpy as np

a = np.array([[4, 5, 3],
              [2, 1, 3],
              [4, 2, 1]])

print("maximum :", a.max())
print("minimum :", a.min())
print("mean :", a.mean())
print("median :", np.median(a))
print("standard deviation :", a.std())
print("determinant :", np.linalg.det(a))
print("rank :", np.linalg.matrix_rank(a))
print("inverse :\n", np.linalg.inv(a))
print("eigen values and vectors :\n", np.linalg.eig(a))
```

```python
import pandas as pd
import numpy as np

data = pd.DataFrame({
    "id": [101, 102, 103, np.nan],
    "name": ["abc", "bch", np.nan, "hst"],
    "age": [np.nan, np.nan, 23, 19],
    "marks": [98, np.nan, 45, 87]
})

print(data)
print(data.tail(2))
print(data.head(3))
print(data.info())
print(data["name"].isna())
print(data["name"].isna().sum())
print(data.fillna(0))
print(data["marks"].mean())
```

```python
import matplotlib.pyplot as plt

name = ["a", "b", "c", "d"]
marks = [87, 78, 98, 56]

plt.bar(name, marks)
plt.xlabel("name")
plt.ylabel("marks")
plt.title("bar graph between name & marks")
plt.show()
```

```python
import matplotlib.pyplot as plt

lang = ["C", "Python", "Java", "C++"]
size = [30, 25, 20, 25]

plt.pie(size, labels=lang, autopct='%1.1f%%')
plt.title("Programming language usage")
plt.show()
```

```python
import matplotlib.pyplot as plt

data = [10, 20, 20, 30, 30, 30, 40, 50, 50]

plt.hist(data, bins=6)
plt.title("Histogram example")
plt.xlabel("values")
plt.ylabel("frequency")
plt.show()
```
